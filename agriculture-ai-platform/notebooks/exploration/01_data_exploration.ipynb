# Data Exploration Notebook

Explorasi data untuk Agriculture AI Platform

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Setup plotting
plt.style.use('seaborn-v0_8')
sns.set_palette('husl')
```

## Load Dataset

```python
# Load metadata
metadata_path = Path('../../data/raw/metadata.csv')
if metadata_path.exists():
    df = pd.read_csv(metadata_path)
    print(f"Dataset shape: {df.shape}")
    df.head()
else:
    print("Metadata file not found. Please prepare the dataset first.")
```

## Class Distribution

```python
if 'df' in locals():
    plt.figure(figsize=(10, 6))
    df['label'].value_counts().plot(kind='bar')
    plt.title('Class Distribution')
    plt.xlabel('Disease Class')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()
```

## Sample Images

```python
from PIL import Image

if 'df' in locals():
    fig, axes = plt.subplots(2, 5, figsize=(15, 6))
    for i, (idx, row) in enumerate(df.sample(10).iterrows()):
        ax = axes[i // 5, i % 5]
        img_path = Path('../../data/raw') / row['image_path']
        if img_path.exists():
            img = Image.open(img_path)
            ax.imshow(img)
            ax.set_title(row['label'])
        ax.axis('off')
    plt.tight_layout()
    plt.show()
```

## Statistical Summary

```python
if 'df' in locals():
    print("Dataset Info:")
    print(df.info())
    print("\nStatistical Summary:")
    print(df.describe())
```
